```python
import os, shutil

d1_location = "/app/training_data/dataset1"
d2_location = "/app/training_data/dataset2"
merged_root = "/app/training_data/merged"

canonical_names = ['ball', 'goalkeeper', 'player', 'referee']
remap_d1 = {0: 1, 1: 2, 2: 0, 3: 2, 4: 3}
remap_d2 = {0: 0, 1: 1, 2: 2, 3: 3}

def apply_remap(location, remap):
    for split in ["train", "valid", "test"]:
        label_dir = f"{location}/{split}/labels"
        if not os.path.isdir(label_dir):
            continue
        for fname in os.listdir(label_dir):
            path = os.path.join(label_dir, fname)
            with open(path) as f:
                lines = f.readlines()
            new_lines = []
            for line in lines:
                parts = line.strip().split()
                if not parts:
                    continue
                old_id = int(parts[0])
                parts[0] = str(remap[old_id])
                new_lines.append(" ".join(parts) + "\n")
            with open(path, "w") as f:
                f.writelines(new_lines)

apply_remap(d1_location, remap_d1)
apply_remap(d2_location, remap_d2)
print("Remapping done.")
```
